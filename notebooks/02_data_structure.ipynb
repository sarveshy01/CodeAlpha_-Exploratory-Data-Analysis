# 2. Data structure

Inspect the shape, types, ranges, missingness, and repeated rows before analysis. Column descriptions are maintained in `data/data_dictionary.md`.

```python
import sys
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(project_root))

from src.data_loader import load_data

df = load_data()
df.info()
```

```python
display(df.describe(include="all").T)
display(df.dtypes.rename("dtype").to_frame())
display(df.nunique(dropna=False).rename("distinct values").to_frame())
```

```python
quality = df.isna().sum().rename("missing").to_frame()
quality["missing_share"] = df.isna().mean()
quality["unique_values"] = df.nunique(dropna=False)
display(quality.sort_values("missing", ascending=False))
print("Exact duplicate rows:", df.duplicated().sum())
```
